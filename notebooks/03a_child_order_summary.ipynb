# 03A — Child Notebook

Called by Notebook 03 using `dbutils.notebook.run()`.


```python
from pyspark.sql import functions as F

dbutils.widgets.text("orders_path", "/Volumes/workspace/retail_training/training_files/official_sources/orders.csv")
dbutils.widgets.text("order_status", "DELIVERED")

orders_path = dbutils.widgets.get("orders_path").strip()
order_status = dbutils.widgets.get("order_status").strip().upper()

if not orders_path:
    raise ValueError("orders_path is required")

valid = {"DELIVERED", "SHIPPED", "PLACED", "CANCELLED"}
if order_status not in valid:
    raise ValueError(f"Unsupported order_status={order_status!r}")
```

```python
orders_df = (
    spark.read.option("header", True).csv(orders_path)
)

processed_count = (
    orders_df
    .filter(
        F.upper(F.trim("order_status")) == F.lit(order_status)
    )
    .count()
)

print("Processed:", processed_count)
```

```python
dbutils.notebook.exit(
    f"order_status={order_status};processed_count={processed_count}"
)
```
